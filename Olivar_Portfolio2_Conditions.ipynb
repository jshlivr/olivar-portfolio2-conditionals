**Introduction** : The objective of this portfolio is to advance the previous Study Reward Calculator by implementing Conditional Execution as discussed in Chapter 3. While the previous version performed linear math, this version utilizes Boolean expressions to evaluate study habits and make decisions. The program now analyzes "Active Days" the frequency of study sessions—using a multi-way (if-elif-else) to determine bonuses or penalties. By using Comparison Operators such as > and ==, the code can differentiate between a "Perfect Streak" and a "Poor Consistency" week. The structure demonstrates the importance of Indentation, where blocks of code are grouped under conditional statements to control the program flow based on user performance.

**Concepts Used:**
- Comparison Operators: I used > to check if hours were recorded for a day and == or >= to categorize the total number of active days into performance tiers.
- Boolean Expressions: The program asks questions like if day > 0 which evaluate to True or False, allowing the computer to decide whether to increment the active day count.
- One-Way Decisions: Inside my loop, I used a simple if statement to add to the active_days counter only when the condition is met, otherwise skipping it.
- Multi-Way Decisions (if-elif-else): This is the core of the reward logic. I used elif to check multiple mutually exclusive conditions (7 days, 5+ days, or 3+ days), ensuring only one bonus or message is applied per week.
- Indentation and Blocks: I increased the indent after every : to define the scope of what happens during a bonus or penalty. I then reduced the indent to return to the main program flow for the final calculation.
- Two-Way Decisions: I utilized an else clause at the end of my logic that applies a penalty if none of the higher-tier consistency goals were met.
- Nested Logic: The program effectively processes data in stages—first gathering and evaluating daily data, then branching based on the result.  

```python
# Constant
rate = 25 

print('--- Advanced Study Reward System ---')

# Input of number of hours for each day 
mon = float(input('Monday: '))
tue = float(input('Tuesday: '))
wed = float(input('Wednesday: '))
thu = float(input('Thursday: '))
fri = float(input('Friday: '))
sat = float(input('Saturday: '))
sun = float(input('Sunday: '))

# 1. Calculate total hours
total_hours = mon + tue + wed + thu + fri + sat + sun
base_reward = total_hours * rate

# 2. Count Active Days (days where study hours > 0)
# We use a list and a simple generator to count
days = [mon, tue, wed, thu, fri, sat, sun]
active_days = 0

for day in days:
    if day > 0:
        active_days += 1

# 3. Multi-tier Logic using if-elif-else
if active_days == 7:
    # Perfect Week: 15% Bonus
    bonus_multiplier = 1.15
    message = "Perfect Streak! 15% Bonus applied."
elif active_days >= 5:
    # Strong Week: 5% Bonus
    bonus_multiplier = 1.05
    message = "Great consistency! 5% Bonus applied."
elif active_days >= 3:
    # Average Week: No Bonus, No Penalty
    bonus_multiplier = 1.0
    message = "Good effort. Standard rates apply."
else:
    # Poor Consistency: 20 Peso Penalty
    bonus_multiplier = 1.0
    base_reward = base_reward - 20
    message = "Penalty applied: Try to study more frequently next week."

# 4. Final Calculation
total_allowance = base_reward * bonus_multiplier

# 5. Output Results
print() 
print('--- Performance Report ---')
print('Total Hours:', total_hours)
print('Active Days:', active_days, '/7')
print('Status:', message)
print('Final Reward: Php', total_allowance)
```
